### 3. Cargar Datos de Clientes (JSON)

### 4. Cargar Datos de Inventario (Excel)

Finalmente, cargaremos los datos de inventario desde una hoja de cálculo de Excel y calcularemos el promedio de stock disponible para todos los productos.

```python
inventario_df = pd.read_excel('/content/drive/MyDrive/Aprendizaje_Automatico/Ejercicio_1/inventario.xlsx')

if 'stock' in inventario_df.columns:
    promedio_stock = inventario_df['stock'].mean()
    print(f"Datos de inventario cargados exitosamente. Promedio de stock disponible: {promedio_stock:.2f}")
else:
    print("La columna 'stock' no fue encontrada en el archivo de inventario. Por favor, verifica el nombre de la columna.")

display(inventario_df.head())
```
